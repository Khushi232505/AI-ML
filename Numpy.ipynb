#***Execution Performance***

```python
size = 10_000_000

#python list
py_list = list(range(size))
start = time.time()
sq_list = [x**2 for x in py_list]
end = time.time()
print(f"python list time = {end-start} seconds")

#np arrays
np_arr = np.array(py_list)
start = time.time()
#vectorization
sq_array = np_arr**2
end = time.time()
print(f"numpy array time = {end-start} seconds")
```

```python
# memory
import sys

print(f"python list size = {sys.getsizeof(py_list) * len(py_list)} bytes")

print(f"numpy array size = {np_arr.nbytes} bytes")
```

#***Create numpy array from list***

```python
# create - from lists

arr = np.array([1, 2, 3, 4, 5])
print(arr, type(arr), arr.dtype, arr.shape)

arr2 = np.array([1, 2, 3, 4, 5, "prime"])
print(arr2, type(arr2), arr2.dtype, arr2.shape)

#2-Dimensional array
arr2D = np.array([[1,2,3], [5,6,7], [8,9,10], [10,11,12]])
print(arr2D, arr2D.shape)
```

#***Using built-in Functions***


```python
arr1 =np.zeros((2,3), dtype = "int64") #prefill with 0
print(arr1, arr1.shape)

arr2 =np.ones((5,4), dtype = "int64") #prefill with 1
print(arr2, arr2.shape)

arr3 =np.full((3,4), 100) #prefill with val
print(arr3, arr3.shape)

arr4 =np.eye((3), dtype = "int64") # identity matrix
print(arr4, arr4.shape)

#range(start, end, step)
arr5 =np.arange(0, 11, 2) # range elements
print(arr5, arr5.shape)

#(start,stop,num)
arr6 =np.linspace(1, 100, 6) #evenly spaced array
print(arr6, arr6.shape)
```
